## 1. Decision Tree Using ID3 Algorithm

A college wants to predict whether a student will **Pass** or **Fail** in an examination based on three categorical attributes:

- **Attendance** – High / Low
- **Study** – Good / Poor
- **Previous Result** – Pass / Fail

### Training Dataset

| Student | Attendance | Study | Previous Result | Final Result |
|:---:|:---:|:---:|:---:|:---:|
| S1 | High | Good | Pass | Pass |
| S2 | High | Good | Fail | Pass |
| S3 | High | Poor | Pass | Pass |
| S4 | High | Poor | Fail | Fail |
| S5 | Low | Good | Pass | Pass |
| S6 | Low | Good | Fail | Fail |
| S7 | Low | Poor | Pass | Fail |
| S8 | Low | Poor | Fail | Fail |
| S9 | High | Good | Pass | Pass |
| S10 | Low | Good | Pass | Pass |
| S11 | High | Poor | Pass | Pass |
| S12 | Low | Poor | Fail | Fail |

### Tasks

1. Calculate the **entropy of the complete dataset** with respect to the `Final Result`.

2. Calculate the **Information Gain** for each attribute:
   - `Attendance`
   - `Study`
   - `Previous Result`

3. Select the attribute having the **highest Information Gain** as the root node.

4. Repeat the **entropy and Information Gain calculations** for the remaining subsets.

5. Construct the complete decision tree using the **ID3 algorithm**.

6. Represent the final decision tree **graphically**.

7. Use the constructed tree to predict the result for the following new students:

| Student | Attendance | Study | Previous Result |
|:---:|:---:|:---:|:---:|
| A | High | Poor | Fail |
| B | Low | Good | Pass |
| C | Low | Poor | Pass |
| D | High | Good | Fail |

8. Implement the same problem using **Scikit-learn's Decision Tree Classifier** and compare the predictions with the manually constructed ID3 tree.


```python
import pandas as pd
import numpy as np
from math import log2
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
data = {
    "Student": ["S1","S2","S3","S4","S5","S6","S7","S8","S9","S10","S11","S12"],
    "Attendance": ["High","High","High","High","Low","Low","Low","Low",
                   "High","Low","High","Low"],
    "Study": ["Good","Good","Poor","Poor","Good","Good","Poor","Poor",
              "Good","Good","Poor","Poor"],
    "Previous Result": ["Pass","Fail","Pass","Fail","Pass","Fail","Pass","Fail",
                        "Pass","Pass","Pass","Fail"],
    "Final Result": ["Pass","Pass","Pass","Fail","Pass","Fail","Fail","Fail",
                     "Pass","Pass","Pass","Fail"]
}

df = pd.DataFrame(data)

print("Training Dataset:")
print(df)

def entropy(y):
    values, counts = np.unique(y, return_counts=True)
    total = len(y)

    ent = 0
    for count in counts:
        p = count / total
        ent -= p * log2(p)

    return ent
def information_gain(df, attribute, target="Final Result"):
    total_entropy = entropy(df[target])

    weighted_entropy = 0

    for value in df[attribute].unique():
        subset = df[df[attribute] == value]
        weighted_entropy += (
            len(subset) / len(df)
        ) * entropy(subset[target])

    return total_entropy - weighted_entropy

print("\nEntropy of Complete Dataset:")
print(round(entropy(df["Final Result"]), 4))
attributes = ["Attendance", "Study", "Previous Result"]

print("\nInformation Gain:")
for attribute in attributes:
    print(
        attribute,
        "=",
        round(information_gain(df, attribute), 4)
    )

def id3(data, attributes, target):
    # If all examples have the same class
    if len(data[target].unique()) == 1:
        return data[target].iloc[0]

    # If no attributes remain
    if len(attributes) == 0:
        return data[target].mode()[0]

    # Find best attribute
    gains = {
        attr: information_gain(data, attr, target)
        for attr in attributes
    }

    best_attribute = max(gains, key=gains.get)

    tree = {best_attribute: {}}

    remaining_attributes = [
        attr for attr in attributes if attr != best_attribute
    ]

    for value in data[best_attribute].unique():
        subset = data[data[best_attribute] == value]

        if len(subset) == 0:
            tree[best_attribute][value] = data[target].mode()[0]
        else:
            tree[best_attribute][value] = id3(
                subset,
                remaining_attributes,
                target
            )

    return tree


manual_tree = id3(
    df,
    ["Attendance", "Study", "Previous Result"],
    "Final Result"
)

print("\nManually Constructed ID3 Decision Tree:")
print(manual_tree)
print("\nSubset Calculations:")

root = max(
    attributes,
    key=lambda x: information_gain(df, x)
)

print("\nRoot Node:", root)

for value in df[root].unique():

    subset = df[df[root] == value]

    print("\nSubset:", root, "=", value)
    print(subset)

    print(
        "Entropy =",
        round(entropy(subset["Final Result"]), 4)
    )

    remaining = [
        attr for attr in attributes if attr != root
    ]

    for attr in remaining:
        print(
            "Information Gain for",
            attr,
            "=",
            round(information_gain(subset, attr), 4)
        )

def predict_manual(tree, sample, default=None):

    if not isinstance(tree, dict):
        return tree

    attribute = next(iter(tree))

    value = sample[attribute]

    if value in tree[attribute]:
        return predict_manual(
            tree[attribute][value],
            sample,
            default
        )

    return default


new_students = pd.DataFrame({
    "Student": ["A", "B", "C", "D"],
    "Attendance": ["High", "Low", "Low", "High"],
    "Study": ["Poor", "Good", "Poor", "Good"],
    "Previous Result": ["Fail", "Pass", "Pass", "Fail"]
})

manual_predictions = []

for _, row in new_students.iterrows():
    prediction = predict_manual(
        manual_tree,
        row.to_dict(),
        df["Final Result"].mode()[0]
    )
    manual_predictions.append(prediction)

new_students["Manual ID3 Prediction"] = manual_predictions

print("\nManual ID3 Predictions:")
print(new_students)

X = pd.get_dummies(
    df[["Attendance", "Study", "Previous Result"]],
    drop_first=False
)

y = df["Final Result"]

model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

model.fit(X, y)

new_X = pd.get_dummies(
    new_students[
        ["Attendance", "Study", "Previous Result"]
    ],
    drop_first=False
)

new_X = new_X.reindex(
    columns=X.columns,
    fill_value=0
)

sklearn_predictions = model.predict(new_X)

new_students["Scikit-learn Prediction"] = sklearn_predictions

print("\nComparison of Predictions:")
print(
    new_students[
        [
            "Student",
            "Manual ID3 Prediction",
            "Scikit-learn Prediction"
        ]
    ]
)

plt.figure(figsize=(14, 8))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True,
    rounded=True
)

plt.title("Decision Tree using Scikit-learn")
plt.show()

print("\nRoot Attribute:", root)
print("\nFinal ID3 Tree:")
print(manual_tree)
```

# 2. Customer Segmentation Using Decision Tree and Online Retail Dataset

Implement a **Decision Tree classifier using the ID3 algorithm** to segment customers based on their purchasing behavior using the **Online Retail dataset**.

## Tasks

1. Load and preprocess the **Online Retail dataset**.

2. Perform the necessary data preprocessing, including:
   - Handling missing values
   - Removing invalid transactions
   - Removing cancelled or returned transactions where appropriate
   - Creating relevant customer-level features
   - Encoding categorical variables where required

3. Implement a **Decision Tree using the ID3 algorithm**.

4. Visualize the constructed decision tree.

5. Analyze the **tree structure**.

6. Calculate and analyze **feature importance**.

7. Discuss how the decision-tree structure helps in understanding **customer purchasing behavior**.

8. Explain which features have the greatest influence on customer segmentation and why.


```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

df=pd.read_excel("Online Retail.xlsx")
df.dropna(subset=["CustomerID"])
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
df['TotalSpend']=df['Quantity']*df['UnitPrice']
customer_df=df.groupby('CustomerID').agg({
    'Quantity': 'sum',
    'UnitPrice' : 'mean',
    'TotalSpend' : 'sum',
    'Country' : 'first'
}).reset_index()

threshold = customer_df['TotalSpend'].median()
customer_df['HighPrice']=(customer_df['TotalSpend']>threshold).astype(int)

le=LabelEncoder()
customer_df['Country']=le.fit_transform(customer_df['Country'])
X=customer_df[['Quantity', 'Country', 'UnitPrice']]
y=customer_df['HighPrice']

model=DecisionTreeClassifier(criterion='entropy', max_depth=4)
model.fit(X,y)
plt.figure(figsize=(20,8))

plot_tree(model,
          feature_names=X.columns,
          class_names=['Low','High'],
          filled=True)

plt.title("Customer Segmentation using Decision Tree (ID3)")
plt.show()
importance = pd.Series(model.feature_importances_, index=X.columns)
print("\nFeature Importance:\n", importance.sort_values(ascending=False))
```

# 3. Logistic Regression vs Decision Tree on Adult Income Dataset

Implement and compare **Logistic Regression** and **Decision Tree** models on the **Adult Income dataset** for predicting income levels.

## Tasks

1. Load and preprocess the **Adult Income dataset**.

2. Perform appropriate preprocessing, including:
   - Handling missing values
   - Encoding categorical features
   - Separating features and target variable
   - Splitting the dataset into training and testing sets
   - Applying feature scaling where required

3. Implement a **Logistic Regression** model.

4. Implement a **Decision Tree** model.

5. Evaluate both models using the following performance metrics:
   - Accuracy
   - Precision
   - Recall
   - F1-score

6. Compare the performance of **Logistic Regression** and **Decision Tree**.

7. Discuss the **interpretability** of both models.

8. Explain which model is more suitable for the **Adult Income dataset** and provide justification based on:
   - Predictive performance
   - Interpretability
   - Feature relationships
   - Model complexity

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score

columns = ['age','workclass','fnlwgt','education','education-num',
           'marital-status','occupation','relationship','race',
           'sex','capital-gain','capital-loss','hours-per-week',
           'native-country','income']
df = pd.read_csv("adult.data", names=columns, na_values=" ?", skipinitialspace=True)
df.dropna(inplace=True)
le = LabelEncoder()
for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col])

X = df.drop('income', axis=1)
y = df['income']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

dt = DecisionTreeClassifier(criterion='entropy', max_depth=5)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

print("=== Logistic Regression ===")
print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print(classification_report(y_test, y_pred_lr))

print("\n=== Decision Tree ===")
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt))
```
