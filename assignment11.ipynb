# 1. K-Means Clustering

## Problem Statement

A shopping mall wants to group its customers into different customer segments based on their Annual Income and Spending Score.

The following data is collected from 12 customers.

| Customer | Annual Income (₹ Lakhs) | Spending Score |
|---|---:|---:|
| C1 | 2.0 | 2.0 |
| C2 | 2.5 | 2.5 |
| C3 | 3.0 | 2.0 |
| C4 | 3.5 | 3.0 |
| C5 | 6.0 | 7.0 |
| C6 | 6.5 | 6.5 |
| C7 | 7.0 | 7.5 |
| C8 | 7.5 | 7.0 |
| C9 | 9.0 | 2.0 |
| C10 | 9.5 | 2.5 |
| C11 | 10.0 | 2.0 |
| C12 | 10.5 | 3.0 |

## Tasks

1. Represent the data using NumPy or Pandas.
2. Plot the customers on a 2-D scatter plot.
3. Apply K-Means with K = 3 and manually demonstrate at least two iterations.
4. Implement K-Means using Scikit-learn and display cluster assignments and centroids.
5. Compare manual and Scikit-learn results.
6. Repeat for K = 2 and K = 4.
7. Calculate inertia for K = 2, 3, 4, and 5, and plot the Elbow Curve.
8. Use the silhouette method to identify a suitable K.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Customer dataset
customers = [
    "C1", "C2", "C3", "C4",
    "C5", "C6", "C7", "C8",
    "C9", "C10", "C11", "C12"
]

X = np.array([
    [2.0, 2.0],
    [2.5, 2.5],
    [3.0, 2.0],
    [3.5, 3.0],
    [6.0, 7.0],
    [6.5, 6.5],
    [7.0, 7.5],
    [7.5, 7.0],
    [9.0, 2.0],
    [9.5, 2.5],
    [10.0, 2.0],
    [10.5, 3.0]
])

df = pd.DataFrame({
    "Customer": customers,
    "Annual Income": X[:, 0],
    "Spending Score": X[:, 1]
})

print(df)

# Initial scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], s=80)

for i, name in enumerate(customers):
    plt.annotate(name, X[i])

plt.xlabel("Annual Income (₹ Lakhs)")
plt.ylabel("Spending Score")
plt.title("Customer Data")
plt.grid()
plt.show()


# --------------------------------------------------
# MANUAL K-MEANS FOR K = 3
# --------------------------------------------------

# Select three initial centroids
centroids = X[[0, 4, 8]].copy()

print("\nInitial Centroids:")
print(centroids)

for iteration in range(10):

    # Calculate Euclidean distances
    distances = np.linalg.norm(
        X[:, np.newaxis, :] - centroids[np.newaxis, :, :],
        axis=2
    )

    # Assign each point to the nearest centroid
    labels = np.argmin(distances, axis=1)

    # Display distances and assignments
    print(f"\nIteration {iteration + 1}")
    print("Distance Matrix:")
    print(np.round(distances, 2))

    print("\nCluster Assignments:")
    for name, label in zip(customers, labels):
        print(name, "-> Cluster", label + 1)

    # Recalculate centroids
    new_centroids = np.array([
        X[labels == k].mean(axis=0)
        if np.any(labels == k)
        else centroids[k]
        for k in range(3)
    ])

    print("\nUpdated Centroids:")
    print(np.round(new_centroids, 2))

    # Stop when centroids no longer change
    if np.allclose(centroids, new_centroids):
        centroids = new_centroids
        print("Clusters have stabilized.")
        break

    centroids = new_centroids


print("\nFinal Manual Results")
for name, label in zip(customers, labels):
    print(name, "-> Cluster", label + 1)

print("\nFinal Manual Centroids:")
print(centroids)


# --------------------------------------------------
# SCATTER PLOT OF MANUAL CLUSTERS
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(
    X[:, 0], X[:, 1],
    c=labels, cmap="viridis", s=100
)

plt.scatter(
    centroids[:, 0], centroids[:, 1],
    marker="X", s=250, color="red",
    label="Centroids"
)

for i, name in enumerate(customers):
    plt.annotate(name, X[i])

plt.xlabel("Annual Income (₹ Lakhs)")
plt.ylabel("Spending Score")
plt.title("Manual K-Means Clustering (K=3)")
plt.legend()
plt.grid()
plt.show()


# --------------------------------------------------
# SCIKIT-LEARN K-MEANS
# --------------------------------------------------

model = KMeans(
    n_clusters=3,
    init=centroids,
    n_init=1,
    random_state=42
)

sk_labels = model.fit_predict(X)

print("\nScikit-learn Cluster Assignments:")
for name, label in zip(customers, sk_labels):
    print(name, "-> Cluster", label + 1)

print("\nScikit-learn Centroids:")
print(model.cluster_centers_)

print("\nManual vs Scikit-learn assignments:")
print(pd.DataFrame({
    "Customer": customers,
    "Manual Cluster": labels + 1,
    "Scikit-learn Cluster": sk_labels + 1
}))


# --------------------------------------------------
# EXPERIMENT WITH K = 2, 3, 4
# --------------------------------------------------

for k in [2, 3, 4]:

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = km.fit_predict(X)

    print(f"\nResults for K = {k}")
    print("Inertia:", round(km.inertia_, 3))

    print(pd.DataFrame({
        "Customer": customers,
        "Cluster": cluster_labels + 1
    }))

    plt.figure(figsize=(7, 5))

    plt.scatter(
        X[:, 0], X[:, 1],
        c=cluster_labels, cmap="viridis", s=100
    )

    plt.scatter(
        km.cluster_centers_[:, 0],
        km.cluster_centers_[:, 1],
        marker="X", s=250, color="red",
        label="Centroids"
    )

    plt.xlabel("Annual Income (₹ Lakhs)")
    plt.ylabel("Spending Score")
    plt.title(f"K-Means Clustering (K={k})")
    plt.legend()
    plt.grid()
    plt.show()


# --------------------------------------------------
# ELBOW METHOD AND SILHOUETTE METHOD
# --------------------------------------------------

inertias = []
silhouette_scores = []

for k in [2, 3, 4, 5]:

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = km.fit_predict(X)

    inertias.append(km.inertia_)
    silhouette_scores.append(
        silhouette_score(X, cluster_labels)
    )

    print(
        f"K={k}, Inertia={km.inertia_:.3f}, "
        f"Silhouette={silhouette_scores[-1]:.3f}"
    )


plt.figure(figsize=(7, 5))
plt.plot([2, 3, 4, 5], inertias, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid()
plt.show()


plt.figure(figsize=(7, 5))
plt.plot([2, 3, 4, 5], silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Method")
plt.grid()
plt.show()

best_k = [2, 3, 4, 5][np.argmax(silhouette_scores)]

print("\nBest K based on silhouette score:", best_k)
```

# 2. Finding the Optimal Number of Clusters Using Elbow and Silhouette Methods

## Problem Statement

Generate a random two-dimensional dataset containing 100 data points and determine the optimal number of clusters using the Elbow Method and Silhouette Analysis.

## Tasks

1. Generate and visualize 100 random 2-D data points.
2. Apply K-Means for K values from 2 to 10.
3. Calculate inertia and plot K vs. Inertia.
4. Calculate silhouette scores and plot K vs. Silhouette Score.
5. Identify the elbow point and the K with the highest silhouette score.
6. Compare both methods and select a suitable K.
7. Visualize the final clusters and their centroids.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Generate random 2-D dataset
np.random.seed(42)
X = np.random.rand(100, 2) * 10

# Visualize the dataset
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1])

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Random 2-D Dataset")
plt.grid()
plt.show()


# Calculate inertia and silhouette scores
k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertias.append(model.inertia_)
    silhouette_scores.append(
        silhouette_score(X, labels)
    )

    print(
        f"K={k}, "
        f"Inertia={model.inertia_:.3f}, "
        f"Silhouette={silhouette_scores[-1]:.3f}"
    )


# Elbow curve
plt.figure(figsize=(7, 5))
plt.plot(list(k_values), inertias, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid()
plt.show()


# Silhouette curve
plt.figure(figsize=(7, 5))
plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Analysis")
plt.grid()
plt.show()


# Select K based on the highest silhouette score
best_k = list(k_values)[np.argmax(silhouette_scores)]

print("\nBest K based on silhouette score:", best_k)


# Final clustering
final_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

final_labels = final_model.fit_predict(X)

print("\nFinal Centroids:")
print(final_model.cluster_centers_)


# Plot final clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X[:, 0], X[:, 1],
    c=final_labels, cmap="viridis", s=60
)

plt.scatter(
    final_model.cluster_centers_[:, 0],
    final_model.cluster_centers_[:, 1],
    marker="X", s=250, color="red",
    label="Centroids"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title(f"Final K-Means Clustering (K={best_k})")
plt.legend()
plt.grid()
plt.show()
```

# 3. Agglomerative Hierarchical Clustering and Dendrogram

## Problem Statement

Use Agglomerative Hierarchical Clustering to group the 12 mall customers based on Annual Income and Spending Score. Visualize the clustering process using a dendrogram.

## Tasks

1. Create the dataset and plot the customers.
2. Calculate the Euclidean distance matrix.
3. Apply Agglomerative Clustering with 3 clusters and single linkage.
4. Display cluster assignments and visualize the clusters.
5. Construct a dendrogram labelled with customer IDs.
6. Draw a horizontal cut line to identify clusters.
7. Compare the dendrogram results with Scikit-learn.
8. Repeat using single, complete, average, and Ward linkage, and compare the results.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.cluster.hierarchy import (
    linkage,
    dendrogram,
    fcluster
)

from scipy.spatial.distance import pdist, squareform

from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import adjusted_rand_score


# Customer dataset
customers = [
    "C1", "C2", "C3", "C4",
    "C5", "C6", "C7", "C8",
    "C9", "C10", "C11", "C12"
]

X = np.array([
    [2.0, 2.0],
    [2.5, 2.5],
    [3.0, 2.0],
    [3.5, 3.0],
    [6.0, 7.0],
    [6.5, 6.5],
    [7.0, 7.5],
    [7.5, 7.0],
    [9.0, 2.0],
    [9.5, 2.5],
    [10.0, 2.0],
    [10.5, 3.0]
])

df = pd.DataFrame({
    "Customer": customers,
    "Annual Income": X[:, 0],
    "Spending Score": X[:, 1]
})

print(df)


# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], s=80)

for i, name in enumerate(customers):
    plt.annotate(name, X[i])

plt.xlabel("Annual Income (₹ Lakhs)")
plt.ylabel("Spending Score")
plt.title("Mall Customer Data")
plt.grid()
plt.show()


# Euclidean distance matrix
distance_matrix = squareform(pdist(X, metric="euclidean"))

print("\nEuclidean Distance Matrix:")
print(pd.DataFrame(
    distance_matrix,
    index=customers,
    columns=customers
).round(2))


# Agglomerative clustering using single linkage
model = AgglomerativeClustering(
    n_clusters=3,
    linkage="single"
)

labels = model.fit_predict(X)

print("\nSingle Linkage Cluster Assignments:")
print(pd.DataFrame({
    "Customer": customers,
    "Cluster": labels + 1
}))


# Plot clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X[:, 0], X[:, 1],
    c=labels, cmap="viridis", s=100
)

for i, name in enumerate(customers):
    plt.annotate(name, X[i])

plt.xlabel("Annual Income (₹ Lakhs)")
plt.ylabel("Spending Score")
plt.title("Agglomerative Clustering - Single Linkage")
plt.grid()
plt.show()


# --------------------------------------------------
# DENDROGRAMS FOR DIFFERENT LINKAGE METHODS
# --------------------------------------------------

linkage_methods = [
    "single",
    "complete",
    "average",
    "ward"
]

linkage_matrices = {}

for method in linkage_methods:

    Z = linkage(X, method=method)
    linkage_matrices[method] = Z

    plt.figure(figsize=(10, 5))

    dendrogram(
        Z,
        labels=customers,
        leaf_rotation=45
    )

    plt.axhline(
        y=4,
        color="red",
        linestyle="--",
        label="Example Cut Line"
    )

    plt.xlabel("Customer")
    plt.ylabel("Linkage Distance")
    plt.title(f"Dendrogram - {method.capitalize()} Linkage")
    plt.legend()
    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# COMPARE LINKAGE METHODS
# --------------------------------------------------

for method in linkage_methods:

    Z = linkage_matrices[method]

    # Obtain three clusters by cutting the dendrogram
    dendrogram_labels = fcluster(
        Z,
        t=3,
        criterion="maxclust"
    )

    # Scikit-learn clustering
    sklearn_model = AgglomerativeClustering(
        n_clusters=3,
        linkage=method
    )

    sklearn_labels = sklearn_model.fit_predict(X)

    print(f"\n{method.capitalize()} Linkage")
    print(pd.DataFrame({
        "Customer": customers,
        "Dendrogram Cluster": dendrogram_labels,
        "Scikit-learn Cluster": sklearn_labels + 1
    }))

    print(
        "Adjusted Rand Index:",
        round(
            adjusted_rand_score(
                dendrogram_labels,
                sklearn_labels
            ),
            3
        )
    )
```

# 4. K-Means Clustering on the Digits Dataset

## Problem Statement

Implement K-Means clustering on the Digits dataset. Experiment with different numbers of clusters and evaluate the results using inertia and silhouette score.

## Tasks

1. Load and preprocess the Digits dataset.
2. Apply K-Means with different values of K.
3. Evaluate clustering using inertia and silhouette score.
4. Analyze how the choice of K affects clustering quality.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA


# Load Digits dataset
digits = load_digits()

X = digits.data
y = digits.target

print("Dataset shape:", X.shape)
print("Number of digit classes:", len(np.unique(y)))


# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# Try different numbers of clusters
k_values = [2, 3, 5, 8, 10, 12]

inertias = []
silhouette_scores = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    inertia = model.inertia_

    # Use a sample for silhouette scoring to reduce computation
    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=min(1000, len(X_scaled)),
        random_state=42
    )

    inertias.append(inertia)
    silhouette_scores.append(score)

    print(
        f"K={k}, "
        f"Inertia={inertia:.2f}, "
        f"Silhouette Score={score:.4f}"
    )


# Inertia plot
plt.figure(figsize=(7, 5))

plt.plot(k_values, inertias, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method - Digits Dataset")
plt.grid()
plt.show()


# Silhouette plot
plt.figure(figsize=(7, 5))

plt.plot(k_values, silhouette_scores, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Analysis - Digits Dataset")
plt.grid()
plt.show()


# Select K with the highest silhouette score
best_k = k_values[np.argmax(silhouette_scores)]

print("\nBest K based on silhouette score:", best_k)


# Train final model
final_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

final_labels = final_model.fit_predict(X_scaled)


# Display cluster distribution
print("\nCluster Distribution:")
print(pd.Series(final_labels).value_counts().sort_index())


# Compare clusters with actual digit labels
comparison = pd.crosstab(
    pd.Series(final_labels, name="Cluster"),
    pd.Series(y, name="Actual Digit")
)

print("\nCluster vs Actual Digit:")
print(comparison)


# Visualize clusters in two dimensions using PCA
pca = PCA(n_components=2)

X_2d = pca.fit_transform(X_scaled)

centroids_2d = pca.transform(final_model.cluster_centers_)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=final_labels,
    cmap="viridis",
    s=15
)

plt.scatter(
    centroids_2d[:, 0],
    centroids_2d[:, 1],
    marker="X",
    color="red",
    s=200,
    label="Centroids"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title(f"Digits Dataset Clustering (K={best_k})")
plt.legend()
plt.grid()
plt.show()
```

# 5. Comparison of Hierarchical and K-Means Clustering on the Mall Customers Dataset

## Problem Statement

Implement and compare Agglomerative Hierarchical Clustering and K-Means on the Mall Customers dataset.

## Tasks

1. Load and preprocess the Mall Customers dataset.
2. Apply K-Means and Agglomerative Clustering.
3. Compare the results using inertia, silhouette score, and visualizations.
4. Discuss the strengths and weaknesses of both clustering methods.

```python
# Assignment 11 - Question 5
# Comparison of K-Means and Agglomerative Clustering
# Dataset: Mall Customers

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# Step 1: Load the uploaded dataset
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print(df.head())

# Step 2: Select features
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

# Step 3: Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Apply K-Means clustering
k = 5

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

df['KMeans Cluster'] = kmeans.fit_predict(X_scaled)

kmeans_inertia = kmeans.inertia_
kmeans_silhouette = silhouette_score(
    X_scaled,
    df['KMeans Cluster']
)

print("\nK-Means Results")
print("Inertia:", round(kmeans_inertia, 3))
print("Silhouette Score:", round(kmeans_silhouette, 3))

# Step 5: Apply Agglomerative Clustering
agg = AgglomerativeClustering(
    n_clusters=k,
    linkage='ward'
)

df['Agglomerative Cluster'] = agg.fit_predict(X_scaled)

agg_silhouette = silhouette_score(
    X_scaled,
    df['Agglomerative Cluster']
)

print("\nAgglomerative Clustering Results")
print("Silhouette Score:", round(agg_silhouette, 3))

# Step 6: Calculate K-Means inertia for different K values
inertias = []

for i in range(2, 8):
    model = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertias.append(model.inertia_)

# Step 7: Plot the elbow curve
plt.figure(figsize=(7, 5))
plt.plot(range(2, 8), inertias, marker='o')
plt.title("K-Means Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(range(2, 8))
plt.grid(True)
plt.show()

# Step 8: Visualize K-Means clusters
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x=X_scaled[:, 0],
    y=X_scaled[:, 1],
    hue=df['KMeans Cluster'],
    palette='viridis',
    s=80
)
plt.title("K-Means Clustering")
plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.show()

# Step 9: Visualize Agglomerative clusters
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x=X_scaled[:, 0],
    y=X_scaled[:, 1],
    hue=df['Agglomerative Cluster'],
    palette='viridis',
    s=80
)
plt.title("Agglomerative Clustering")
plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.show()

# Step 10: Compare both algorithms
comparison = pd.DataFrame({
    'Algorithm': ['K-Means', 'Agglomerative Clustering'],
    'Inertia': [round(kmeans_inertia, 3), 'Not directly available'],
    'Silhouette Score': [
        round(kmeans_silhouette, 3),
        round(agg_silhouette, 3)
    ]
})

print("\nAlgorithm Comparison:")
print(comparison.to_string(index=False))

# Step 11: Display cluster assignments
print("\nCustomer Cluster Assignments:")
print(df.to_string(index=False))
```
